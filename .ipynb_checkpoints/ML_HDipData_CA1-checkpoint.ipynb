# Converting categorical columns from string to numeric

```python
cols_to_numeric = [
    "Marital status",
    "Nationality",
    "Father's occupation",
    "Displaced",
    "Tuition fees up to date",
    "Age at enrollment",
    "Curricular units 1st sem (credited)"
]
```

```python
for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.info()
```

```python
df.describe()
```

# Handling Target label

```python
df["Target"].unique()
```

```python
df['Target'].value_counts()
```

# Converting the Target label into three categories

```python
# The text was cleaned using str.strip() to remove whitespace and str.lower() to convert all labels to lowercase

df['Target'] = df['Target'].str.strip().str.lower()

df['Target'] = df['Target'].replace({
    'graduate': 'Graduate',
    'grad': 'Graduate',
    'dropout': 'Dropout',
    'drop': 'Dropout',
    'enrolled': 'Enrolled'
})
```

```python
df['Target'].value_counts()
```

```python
import sklearn.preprocessing as preproc

le=preproc.LabelEncoder()
df["Target"] = le.fit_transform(df["Target"])
```

```python
df["Target"].unique()
```

```python
df['Target'].value_counts()
```

```python
df.head()
```

# Feature Selection
Feature selection was based on relevance to student outcomes, avoiding data leakage, and removing weak or noisy predictors, resulting in the strongest predictors of enrollment, dropout, or graduation.

```python
cols_drop = [
    "Course",
    "Daytime/evening attendance",
    "Mother's occupation",
    "Father's occupation",
    "Mother's qualification",
    "Father's qualification"
]

df = df.drop(columns=cols_drop)
```

```python
df.head()
```

# Visualisation

# Target Label

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Target', data=df)
plt.title("Distribution of Target Classes")
plt.show()
```

# Demographic 

```python
fig, axes = plt.subplots(2, 3, figsize=(18,12))

sns.countplot(x='Gender', data=df, ax=axes[0,0])
axes[0,0].set_title("Gender Distribution")
axes[0,0].tick_params(axis='x', rotation=0)

sns.countplot(x='Marital status', data=df, ax=axes[0,1])
axes[0,1].set_title("Marital Status")
axes[0,1].tick_params(axis='x', rotation=45)

sns.countplot(x='Nationality', data=df, ax=axes[0,2])
axes[0,2].set_title("Nationality")
axes[0,2].tick_params(axis='x', rotation=45)

sns.countplot(x='Scholarship holder', data=df, ax=axes[1,0])
axes[1,0].set_title("Scholarship Holder")
axes[1,0].tick_params(axis='x', rotation=45)

sns.countplot(x='Displaced', data=df, ax=axes[1,1])
axes[1,1].set_title("Displaced")
axes[1,1].tick_params(axis='x', rotation=45)

sns.countplot(x='International', data=df, ax=axes[1,2])
axes[1,2].set_title("International Student")
axes[1,2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

# Academic Performance 

```python
grade_cols = [
    'Previous qualification (grade)',
    'Admission grade',
    'Curricular units 1st sem (grade)',
    'Curricular units 2nd sem (grade)'
]

df[grade_cols].hist(figsize=(12,8))
plt.suptitle("Grade Distributions")
plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(18, 14), dpi=120)

corr = df.corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", annot_kws={"size":8})

plt.title("Correlation Heatmap", fontsize=18)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.tight_layout()
plt.show()
```

All the visualisations were used to understand the dataset better. They show how students are distributed across categories, how their grades vary, and how the variables relate. These plots help identify important patterns, support decisions like using the mode or median for missing values, and show why academic performance is the strongest predictor of whether a student graduates, drops out, or enrolls.
