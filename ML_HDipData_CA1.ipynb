# Converting categorical columns from string to numeric

```python
cols_to_numeric = [
    "Marital status",
    "Nationality",
    "Father's occupation",
    "Displaced",
    "Tuition fees up to date",
    "Age at enrollment",
    "Curricular units 1st sem (credited)"
]
```

```python
for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```
