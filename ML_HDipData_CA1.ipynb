# Three training splits (5%, 15% and 25%) using 10-fold cross validation techniques with RandomForest, SVM and LogisticRegression

```python
def cv10(model, X_train, y_train):
    return cross_val_score(model, X_train, y_train, cv=10, scoring="accuracy").mean()
```

# 5% SPLIT — RF vs SVM vs LR

```python
# Random Forest 

rf_5 = RandomForestClassifier(random_state=102)
rf_5.fit(X_train_5, y_train_5)
y_pred_rf_5 = rf_5.predict(X_test_5)

print(" Random Forest: Classification Report (5%) ")
print(classification_report(y_test_5, y_pred_rf_5))

print(" Random Forest: Confusion Matrix (5%) ")
print(confusion_matrix(y_test_5, y_pred_rf_5))

rf_cv_5 = cv10(RandomForestClassifier(random_state=102), X_train_5, y_train_5)
print(f"\nRandom Forest Mean CV Accuracy (10 folds, 5%): {rf_cv_5:.4f}")
```

```python
# SCALE DATA for SVM + LR

scaler_5 = StandardScaler()
X_train_5_scaled = scaler_5.fit_transform(X_train_5)
X_test_5_scaled = scaler_5.transform(X_test_5)
```

```python
# SVM (scaled)

svm_5 = SVC()
svm_5.fit(X_train_5_scaled, y_train_5)
y_pred_svm_5 = svm_5.predict(X_test_5_scaled)

print(" SVM: Classification Report (5%) ")
print(classification_report(y_test_5, y_pred_svm_5))

print(" SVM: Confusion Matrix (5%) ")
print(confusion_matrix(y_test_5, y_pred_svm_5))

svm_cv_5 = cv10(SVC(), X_train_5_scaled, y_train_5)
print(f"\nSVM Mean CV Accuracy (10 folds, 5%): {svm_cv_5:.4f}")
```

```python
# Logistic Regression (scaled)

lr_5 = LogisticRegression(max_iter=2000, random_state=102)
lr_5.fit(X_train_5_scaled, y_train_5)
y_pred_lr_5 = lr_5.predict(X_test_5_scaled)

print(" Logistic Regression: Classification Report (5%) ")
print(classification_report(y_test_5, y_pred_lr_5))

print(" Logistic Regression: Confusion Matrix (5%) ")
print(confusion_matrix(y_test_5, y_pred_lr_5))

lr_cv_5 = cv10(LogisticRegression(max_iter=2000, random_state=102), X_train_5_scaled, y_train_5)
print(f"\nLogistic Regression Mean CV Accuracy (10 folds, 5%): {lr_cv_5:.4f}")
```

```python
from sklearn.pipeline import Pipeline
splits = {
    "5%": train_test_split(X, y, test_size=0.05, stratify=y, random_state=42),
    "15%": train_test_split(X, y, test_size=0.15, stratify=y, random_state=42),
    "25%": train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)
}

results_splits = []

for split_name, (X_train, X_test, y_train, y_test) in splits.items():
    # LR
    lr_pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(max_iter=500, random_state=42))
    ])
    lr_pipe.fit(X_train, y_train)
    lr_acc = accuracy_score(y_test, lr_pipe.predict(X_test))

    # RF
    rf = RandomForestClassifier(random_state=42)
    rf.fit(X_train, y_train)
    rf_acc = accuracy_score(y_test, rf.predict(X_test))

    # SVM
    svm_pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("clf", SVC())
    ])
    svm_pipe.fit(X_train, y_train)
    svm_acc = accuracy_score(y_test, svm_pipe.predict(X_test))

    results_splits.append([split_name, lr_acc, rf_acc, svm_acc])

results_splits_df = pd.DataFrame(
    results_splits,
    columns=["Split", "LR Accuracy", "RF Accuracy", "SVM Accuracy"]
)
print(results_splits_df)
```
